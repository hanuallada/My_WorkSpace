```python
new_tuple1=(10,20,10,10,20,30,40,30,50,60,70,80,90,100)
print(new_tuple1)
print(len(new_tuple1)) #prints the length of the tuple
print(new_tuple1.count(10)) #prints the number of occurrences of 10 in the tuple
print(new_tuple1.index(20)) #prints the index of the first occurrence of 20 in the tuple
print(new_tuple1.index(30)) #prints the index of the first occurrence of 30 in the tuple
print(new_tuple1.index(30, 4)) #prints the index of the first occurrence of 30 in the tuple starting from index 4
print(new_tuple1.index(30, 4, 10)) #prints the index of the first occurrence of 30 in the tuple starting from index 4 and ending at index 10
print(new_tuple1.index("Hanumanth")) #gives error because "Hanumanth" is not present in the tuple
```

```python
# Tuple concatenation
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
tuple3 = tuple1 + tuple2
print(tuple3)  # Output: (1, 2, 3, 4, 5, 6)
```

```python
fruits = ("apple", "banana", "cherry", "date", "elderberry")
sub_fruits = fruits[1:4]  # Slicing from index 1 to 3 (4 is not included)
print(sub_fruits)  # Output: ('banana', 'cherry', 'date')  
```

```python
tuple1=('a','b','c','d','e','f','g')
print(tuple1)
print(tuple1[-3]) #prints the third last element of the tuple
print(tuple1[2:5]) #prints the elements from index 2 to 4
print(tuple1[2:]) #prints the elements from index 2 to the end
print(tuple1[:5]) #prints the elements from the start to index 4
```
