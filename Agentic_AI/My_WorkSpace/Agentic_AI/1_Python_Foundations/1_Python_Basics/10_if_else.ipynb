```python
acquired_marks = 65

if acquired_marks >= 90:
    print("Topper")
elif acquired_marks >= 80 and acquired_marks < 90:
    print("Second Class")
elif acquired_marks >= 70 and acquired_marks < 80:
    print("Third Class")
elif acquired_marks >= 60 and acquired_marks < 70:
    print("Fourth Class")
elif acquired_marks >= 50 and acquired_marks < 60:
    print("Fifth Class")
elif acquired_marks >= 35 and acquired_marks < 50:
    print("Just Passed")
else:
    print("Fail")
```
