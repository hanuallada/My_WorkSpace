```python
# External Data type casting in Python
x = 5  # integer
y = 2.5  # float
z = "10"  # string
print("Original data types:")
print("x:", type(x))
print("y:", type(y))
print("z:", type(z))
print("\nCasting data types:")
print(x + y + int(z))  # casting string to integer
print(type(x + y + int(z)))
```

```python
# Internal Type Conversion
a = 10  # integer
b = 3.5  # float
result = a + b  # automatic type conversion to float
print("\nInternal Type Conversion:")
print("a:", type(a))
print("b:", type(b))
print("Result of a + b:", result)
```

```python
# convert the values which are compatible only
print(int('10a'))  # This will raise a ValueError because '10a' cannot be converted to an integer
```

```python
emp_name="Hanumantha Rao Allada"
print(tuple(emp_name))  # This will raise a TypeError because a string cannot be directly converted to a tuple
print(list(emp_name))  # This will raise a TypeError because a string cannot be directly converted to a list
print(set(emp_name))  # This will raise a TypeError because a string cannot be directly converted to a set
```

```python
# list to tuple conversion
my_list = [1, 2, 3, 4, 5]
my_tuple = tuple(my_list)
print("Original list:", my_list)
print("Converted tuple:", my_tuple)
```

```python
#tuple to list conversion
my_tuple = (1, 2, 3, 4, 5)
my_list = list(my_tuple)
print("Original tuple:", my_tuple)
print("Converted list:", my_list)
```

```python
a=1
b=2
c=complex(a,b)
print("Complex number:", c)
print("Real part:", c.real)
print(type(c.real))
print("Imaginary part:", c.imag)
print(type(c.imag))
print("Type of complex number:", type(c))
```
