```python
fruits=["apple","banana","cherry"] #Lists conatining frutis
print(fruits)
print(type(fruits))
```

```python
for i in fruits:
    print(fruits)
```

```python
for i in fruits:
    print(i)
```

```python
# After seeing the above two examples. we can clearly see that below understandings
# for --> keyword indicating the start of the loop
# Iterable --> The sequence or Iterable object being looped for. Example here fruits list is iterable in above example
# variable/Iterator --> It takes on the value of each item in the iterable during each iteration.
```

```python
# Range Function
# We can generate a sequence of numbers using range() function.
# Range(10) generates numbers from 0 to 9 (10 numbers).
# We can also define the start, stop and step size as well as per our needs.
# default step size is 1 if we don't specify.
# Range fucntions generally doesn't store all the values in memory as it would be inefficient.
# so it remembers the start, stop and step size and generates the next number on the go.
```

```python
for i in range(5):
    print(i)
```

```python
for i in range(0,20,5):
    print(i)
```

```python
# Using length and range together instead of index values to fetch iterable values using for loop
print(fruits)
for i in range(len(fruits)):       # here len gives 3 and range(3) gives 0,1,2 and we can use those values as index in each iteration
    print(fruits[i])
```

```python
# Enumerate merges direct iterator in for loop and above length and range combo together
# Enumerator generates two items one is index and other is value as Index, Value
print(fruits)
print(enumerate(fruits))        # returns Iterbale holding result as Index, Value
for index, value in enumerate(fruits):
    print("At position {}, we have fruit named {}".format(index,value))
```

```python
# if else condition inside a for loop
nums = [1,2,3,4,5,6]

for n in nums:
    if n%2 == 0:
        print("Number {} is even number".format(n))
    else:
        print("Number {} is odd number".format(n))
```

```python
students=[{"name":"Hanu","Percentage":90},
          {"name":"Allada","Percentage":80},
          {"name":"Rao","Percentage":70},
          {"name":"Hanumanth", "Percentage":30}]
print(students)
print(type(students))
print(type(students[0]))
```

```python
# Combination of dictionaries, list, for loop and if else if
for stud in students:
    stud_name=stud["name"]
    stud_Perc=stud["Percentage"]

    if stud_Perc >= 90:
        print(f"{stud_name}% is {stud_Perc} and he is excellent")
    elif stud_Perc >=80 and stud_Perc < 90:
        print(f"{stud_name} % is {stud_Perc} and he is Good")
    elif stud_Perc >= 70 and stud_Perc < 80:
        print(f"{stud_name} % is {stud_Perc} and he is Average")
    else:
        print(f"{stud_name} % is {stud_Perc} and he is Poor")
```

```python
# Prime numbers falls between given two range of numbers
start_num = -3
end_num = 50

for num in range(start_num, end_num+1):  # to include the end_num, added +1 to end_num as range considers before the end_num
    if num > 1:
        prime_cnt=0
        for i in range(2,num):
            if num%i==0:
                prime_cnt += 1
                break # we can keep this to stop if its already divided by other nums to save some iterations/processing time.
        if prime_cnt == 0:
            print(f"{num} is Prime Number")
    else:
        print(f"{num} is negative number or less than Prime minimum value")

```
