```python
mystring='python'
print(mystring[0]) # prints first character
print(mystring[1]) # prints second character
print(mystring[1:4]) # prints characters starting from 2nd to 4th
print(mystring[2:]) # prints characters starting from 3rd character
```

```python
string1='Hello'
string2='World'
print(string1 + ' ' + string2) # prints Hello World
concat_string=string1 + ' ' + string2
print(concat_string) # prints Hello World
```

```python
string3=10
concat_string=string1 + ' ' + string2 + ' ' + string3
print(concat_string)
```

```python
string3=10
concat_string=string1 + ' ' + string2 + ' ' + str(string3)
print(concat_string)
```
