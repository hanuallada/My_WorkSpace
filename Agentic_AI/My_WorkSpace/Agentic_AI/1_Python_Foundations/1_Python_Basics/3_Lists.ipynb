```python
#List within a list or nested list
mylist5 = [[1, 2], [3, 4,5], [6,7,8,9], ['Hanu',37,'200k']]
print(mylist5)
```

```python
#fetching an element from a list within a list
print(mylist5[1][2]) #fetching 3rd element from 2nd list
print(mylist5[3][0]) #fetching 1st element from 4th list
```

```python
# List element manipulation
print(mylist2)
mylist2.append(7) #Appends 7 to the end of the list
mylist2.insert(2,10) #Inserts 10 at index 2
mylist2.pop(1) #Removes the element at index 1
mylist2.remove('a') #Removes the first occurrence of '3.5'
print(mylist2)
```

```python
# List comprehension
squares = [x**2 for x in range(10)]
print(squares)
```
