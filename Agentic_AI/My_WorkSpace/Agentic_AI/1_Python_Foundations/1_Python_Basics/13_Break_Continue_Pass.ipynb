```python
# Break keyword breaks out/stops the current loop

numbers= [1,2,3,4,5];

for i in numbers:
    if i == 3:
        print("Found the number 3")
        break
    print(i)
```

```python
# For else

numbers= [1,2,3,4,5];

for i in numbers:
    if i == 3:
        print("Found the number 3")
        break
    print(f"current number is {i}")
else:
    print("stil insde the loop")
  
print("Came out of the loop")
```

```python
# Conitnue is used to skip that current iteration but will continue with next iterable values
# SKipping even numbers and printing only odd numbers

numbers= [1,2,3,4,5]

for i in numbers:
    if i%2 == 0:
        print(f"Skipping the number {i}")
        continue
    print(f"Printing the number {i}")
```

```python
# Pass is used to do nothing

numbers = [1,2,3,4,5,6]
for i in numbers:
    if i == 4:
        print(f"Skipping the number {i}")
        pass
    else:
        print(i)
```
