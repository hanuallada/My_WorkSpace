```python
# List of Arithematic Operators in Python
# "+" for Addition
# "-" for subtraction
# "*" for multiplication
# "/" for division
# "%" for Modulus to get remainder of the division
# "**" for exponentiation
# "//" for floor division - rounded to nearest whole number
```

```python
a= 10
b= 3
print(a+b)  # Addition
print(a-b)  # Subtraction
print(a*b)  # Multiplication
print(a/b)  # Division
print(a%b)  # Reminder of the division
print(a**b) # 2**2 = 4, 3**2 = 9
print(a//b) # 10/3=3.33 and floor value of 3.33 is 3
```

```python
# List of Comparison Operators in Python
# 1. == (equals to)
# 2. != (Not equals to)
# 3. <  (Less than)
# 4. >  (Greater than)
# 5. <= (Less than or equals to)
# 6. >= (Greater than or equals to)
```

```python
x=5
y=10
print((x==y))
print((x!=y))
print((x<y))
print((x>y))
print((x<=y))
print((x>=y))
```

```python
# Logical Operators in Python
# 1. and (Logical AND)
# 2. or  (Logical OR)
# 3. not (Logical NOT)
```

```python
a=True
b=False
print(a and b)
print(a or b)
print(not b)
```

```python
x=5
y=3
print(x<10 or y > 10)
```

```python
# Asignment Operator
# = represents assignment and it assigns the value on the right to the variable on the left
# +=, -=, *=, /=, %= are compund assignment operators
a=10
b=3

c = a   # simple assignment
b += 5  # Additional assignment and equivalent to b = b+5
a -= 2  # Subtraction assignment and equivalent to a = a-2
c *= 3  # Multiplication assignment and equivalent to c = c*3
b /= 2  # Division assignment and equivalent to b = b/2
a %= 4  # Modulus assignment and equivalane to a = a%4
c //= 2 # Floor Division assignment and equivalane to c = C//2
b **= 2 # Exponentiation assignment and equiavalent to b = b**2
print("a = ",a)
print("b = ",b)
print("c = ",c)
```

```python
# Special Operators
# is and is not are the identity operators in python
# They are used to check whether two variables/values are located on the same part of the memory.

a = 5
b = 5
print( a is b)

list1 = [1,2,3,4,5]
list2 = [1,2,3,4,5]
print(list1 is list2)

name1="Hanu"
name2="Hanu"
print(name1 is name2)
```

```python
print(a,b)
print(id(a), id(b))
print(list1,list2)
print(id(list1), id(list2))
print(name1, name2)
print(id(name1), id(name2))
```
