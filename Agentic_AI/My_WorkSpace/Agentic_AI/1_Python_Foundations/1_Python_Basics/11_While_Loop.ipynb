```python
# To perform a taks multiple time for a given set of inpts or entries then we will go for loops (While loop, For loop)
# In While loop, the body will be continuously executed until the condition is true
# Iteration over each input/entry should be stopped at one point either by updating the iterator in the loop or condition failure at one point

counter = 1
while counter < 10:
    counter += 1
    print("The Number updated to ",counter)
```

```python
# Nested while loop example
teacher_cnt = 1
principal_cnt = 1

while teacher_cnt <= 3:
    print("Teacher's Turn ",teacher_cnt)
    while principal_cnt <= 3:
        print("--> Principal's Turn ",principal_cnt)
        principal_cnt += 1
    teacher_cnt += 1
```

```python
num_list = [2,3,5,7,10]
sum_sqaure_nums= 0
index=0

while index < len(num_list):
    sum_sqaure_nums += num_list[index]**2
    index += 1

print("Sum of squares of all given numbers is ", sum_sqaure_nums)
```

```python
# While loop Else

teacher_cnt = 1
principal_cnt = 1

while teacher_cnt <= 3:
    print("Teacher's Turn ",teacher_cnt)
    while principal_cnt <= 3:
        print("--> Principal's Turn ",principal_cnt)
        principal_cnt += 1
    else:
        print("--> Principal's Turn already completed")
    teacher_cnt += 1
```
